```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

sns.set_theme(style='whitegrid')

sinasc = pd.read_csv('/content/SINASC_RO_2019_JAN.csv')

def plota_pivot_table(df, value, index, func, ylabel, xlabel, opcao='nada'):
    if opcao == 'nada':
        pd.pivot_table(df, values=value, index=index,
                       aggfunc=func).plot(figsize=[15, 5])
    elif opcao == 'sort':
        pd.pivot_table(df, values=value, index=index,
                       aggfunc=func).sort_values(value).plot(figsize=[15, 5])
    elif opcao == 'unstack':
        pd.pivot_table(df, values=value, index=index,
                       aggfunc=func).unstack().plot(figsize=[15, 5])
    plt.ylabel(ylabel)
    plt.xlabel(xlabel)
    return None

max_data = sinasc.DTNASC.max()[:7]
max_data

os.makedirs('./output/figs/'+max_data, exist_ok=True)

# 1 - Quantidade de nascimentos/Sexo
plota_pivot_table(sinasc, 'IDADEMAE', ['DTNASC', 'SEXO'], 'count', 'quantidade de nascimento','data de nascimento', 'unstack')
plt.savefig('./output/figs/'+max_data+'/quantidade de nascimento por data.png')

# 2 - Média Idade da Mãe/Sexo
plota_pivot_table(sinasc, 'IDADEMAE', ['DTNASC', 'SEXO'], 'mean', 'media idade mae','data de nascimento','unstack')
plt.savefig('./output/figs/'+max_data+'/media idade mae por sexo.png')

# 3 - Média Peso Bebe/Sexo
plota_pivot_table(sinasc, 'PESO', ['DTNASC', 'SEXO'], 'mean', 'media peso bebe','data de nascimento','unstack')
plt.savefig('./output/figs/'+max_data+'/media peso bebe por sexo.png')

# 4 - Média peso/Escolaridade Mãe
plota_pivot_table(sinasc, 'PESO', 'ESCMAE', 'median', 'peso','escolaridade mãe','sort')
plt.savefig('./output/figs/'+max_data+'/media peso por escolaridade mae.png')

# 5 - Apgar1/Gestação
plota_pivot_table(sinasc, 'APGAR1', 'GESTACAO', 'mean', 'apgar1 medio','gestacao','sort')
plt.savefig('./output/figs/'+max_data+'/media apgar1 por gestacao.png')

# 6 - Apgar5/Gestação
plota_pivot_table(sinasc, 'APGAR5', 'GESTACAO', 'mean', 'apgar5 medio','gestacao','sort')
plt.savefig('./output/figs/'+max_data+'/media apgar5 por gestacao.png')
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

def generate_plots_for_month(month_abbr):
    """
    Generates all required plots for a given month's data.

    Args:
        month_abbr (str): The three-letter abbreviation of the month (e.g., 'JAN', 'FEB').
    """
    file_path = f'/content/SINASC_RO_2019_{month_abbr}.csv'

    # Check if the file exists before processing
    if not os.path.exists(file_path):
        print(f"Warning: CSV file not found for {month_abbr}: {file_path}. Skipping.")
        return

    sinasc_month = pd.read_csv(file_path)

    # Convert 'DTNASC' to datetime and extract max_data_month
    # Assuming DTNASC is in 'YYYY-MM-DD' format
    sinasc_month['DTNASC'] = pd.to_datetime(sinasc_month['DTNASC'])
    max_data_month = sinasc_month['DTNASC'].max().strftime('%Y-%m')

    output_dir = f'./output/figs/{max_data_month}'
    os.makedirs(output_dir, exist_ok=True)

    print(f"Generating plots for {month_abbr} (data up to {max_data_month})...")

    # 1 - Quantidade de nascimentos/Sexo
    plota_pivot_table(sinasc_month, 'IDADEMAE', ['DTNASC', 'SEXO'], 'count', 'quantidade de nascimento','data de nascimento', 'unstack')
    plt.savefig(f'{output_dir}/quantidade de nascimento por data.png')
    plt.close() # Close plot to free memory

    # 2 - Média Idade da Mãe/Sexo
    plota_pivot_table(sinasc_month, 'IDADEMAE', ['DTNASC', 'SEXO'], 'mean', 'media idade mae','data de nascimento','unstack')
    plt.savefig(f'{output_dir}/media idade mae por sexo.png')
    plt.close()

    # 3 - Média Peso Bebe/Sexo
    plota_pivot_table(sinasc_month, 'PESO', ['DTNASC', 'SEXO'], 'mean', 'media peso bebe','data de nascimento','unstack')
    plt.savefig(f'{output_dir}/media peso bebe por sexo.png')
    plt.close()

    # 4 - Média peso/Escolaridade Mãe
    plota_pivot_table(sinasc_month, 'PESO', 'ESCMAE', 'median', 'peso','escolaridade mãe','sort')
    plt.savefig(f'{output_dir}/media peso por escolaridade mae.png')
    plt.close()

    # 5 - Apgar1/Gestação
    plota_pivot_table(sinasc_month, 'APGAR1', 'GESTACAO', 'mean', 'apgar1 medio','gestacao','sort')
    plt.savefig(f'{output_dir}/media apgar1 por gestacao.png')
    plt.close()

    # 6 - Apgar5/Gestação
    plota_pivot_table(sinasc_month, 'APGAR5', 'GESTACAO', 'mean', 'apgar5 medio','gestacao','sort')
    plt.savefig(f'{output_dir}/media apgar5 por gestacao.png')
    plt.close()

    print(f"Plots for {month_abbr} generated successfully.")

# List of month abbreviations
months = ['JAN', 'FEV', 'MAR', 'ABR', 'MAI', 'JUN', 'JUL', 'AGO', 'SET', 'OUT', 'NOV', 'DEZ']

for month in months:
    generate_plots_for_month(month)

print("All monthly plots generated (if CSV files were found).")
```
